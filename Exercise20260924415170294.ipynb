1. 哪些是 Python 的內建型別（Built-in Types）？
Python 內建了多種核心資料型別，不需要額外引入模組即可直接使用，主要包含以下幾大類：

數值型別：整數 (int)、浮點數 (float)、複數 (complex)。

序列型別：字串 (str)、列表 (list)、元組 (tuple)、範圍 (range)。

對應與集合：字典 (dict)、集合 (set)、凍結集合 (frozenset)。

其他基本型別：布林值 (bool)、空值 (NoneType)。
2. 哪些需要額外模組才能較完整地表示？
雖然內建型別功能強大，但若要進行精確的科學計算或特殊資料表示，通常需要搭配標準庫或第三方模組：

高精度十進位與分數：需使用 decimal 模組（避免浮點數誤差）與 fractions 模組（精確表示分數）。

矩陣與多維陣列：需使用第三方 numpy 模組，才能完整對應數學上的向量與矩陣空間。

進階數學運算：需使用 math 或 cmath 模組來處理三角函數、對數與複數的高階運算。

時間與日期：需使用 datetime 模組來精確表示年、月、日、時區與時間間隔。
3. Python 的型別分類與數學中的分類是否完全一致？
Python 的型別分類與數學上的抽象分類並不完全一致，主要差異在於電腦硬體的限制：連續性與離散性：數學上的實數是連續且無限的；而 Python 的 float 是基於 IEEE 754 標準的有限位元浮點數，屬於離散且有精度極限的近似值。整數大小：數學上的整數無限大，Python 的 int 雖然支援動態擴充位元以支援任意大小整數，但仍受限於實體記憶體總量。運算結果型別：數學中兩整數相除（例如 1/ 2）在代數上是有理數，但在 Python 3 的標準除法中會直接轉換成浮點數。

請回答：

1. Python 是否為上述每一個數系都提供獨立型別？
不是的。 數學上的數系非常細緻（包含自然數、整數、有理數、無理數、實數、複數等），但 Python 基於實用性、記憶體效率與語言簡潔度，並沒有為每一個數學數系都設立一個獨立的頂級內建型別。它只挑選了最核心、最常用的幾種（如 int、float、complex），其他則透過標準庫或第三方庫來補充。
2. 為什麼自然數與整數都使用 `int` 表示？
計算與記憶體效率：在電腦科學中，處理整數的硬體邏輯是通用的。Python 的 int 具備自動擴充位元的大小（可以處理任意大的整數），因此非負的自然數與負整數在底層共用同一套機制是最有效率的做法。

語言設計簡化：如果硬性將「自然數」獨立成一個型別（例如 Natural），開發者在進行減法或賦值時就會面臨繁瑣的型別限制與額外轉型，這違反了 Python 追求直覺與實用的初衷。
3. Python 為什麼提供 `complex` 型別？
支援科學與工程計算：Python 長期被廣泛應用於科學計算、資料科學、電機工程（如交流電分析）以及量子力學等領域。語法直覺：在這些領域中，複數（$a + bi$）是核心工具。Python 直接內建了 complex 型別（語法上使用 j 或 J 表示虛部，例如 3 + 4j），讓科學家和工程師可以直接對複數進行數學運算，而不必自行寫類別去管理實部與虛部。
4. Python 為什麼沒有內建的有理數型別？
嚴格來說，Python 其實有提供有理數支援，只是它沒有被放在最核心的全域內建型別中，而是放在標準庫的 fractions 模組（透過 Fraction 類別）裡。

效能與輕量化考量：有理數（分數）在運算時需要不斷進行分子分母的互質化簡與大數運算，這會消耗較多的運算效能。

多數場景不需要：絕大多數日常程式開發與硬體層面的數值計算，利用速度極快的二進位浮點數（float）就能完美應付。因此，Python 選擇將有理數作為「按需引入」的標準庫模組，保持核心語言的輕量與高效。
5. 下列三個物件的型別有何不同？
這三個物件在 Python 中屬於三種完全不同的數值型別
```python
1
1.0
1+0j
```

## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？
2. 對應的 CPython 物件結構名稱為何？
3. 它如何儲存資料？
4. 是否有固定大小限制？
5. 若資料規模持續增加，如何擴充儲存空間？
1. int（整數 / 自然數）
是否為內建型別：是。

CPython 物件結構名稱：PyLongObject。

如何儲存資料：使用變動長度的陣列（ob_digit）來儲存大整數的數位，並透過標頭記錄正負號與長度。

是否有固定大小限制：無硬性大小限制（支援任意精度運算，Arbitrary Precision）。

如何擴充儲存空間：當數值超過當前配置的位數時，CPython 會自動向系統申請更大的記憶體空間來擴充 ob_digit 陣列。

2. float（實數）
是否為內建型別：是。

CPython 物件結構名稱：PyFloatObject。

如何儲存資料：內部直接包裝並儲存一個 C 語言的雙精度浮點數（double）。

是否有固定大小限制：有，固定佔用 8 位元組（64 bits），精確度與範圍受 IEEE 754 標準限制。

如何擴充儲存空間：不需要擴充，因為其記憶體大小與結構在建立時即完全固定。

3. complex（複數）
是否為內建型別：是。

CPython 物件結構名稱：PyComplexObject。

如何儲存資料：內部包含兩個 C 語言的 double，分別用來儲存實部（real）與虛部（imag）。

是否有固定大小限制：有，固定大小（包含兩個 double 與物件標頭）。

如何擴充儲存空間：不需要擴充，記憶體大小固定不變。

4. bool（命題真假）
是否為內建型別：是。

CPython 物件結構名稱：PyBoolObject（實際上是 PyLongObject 的子類別）。

如何儲存資料：採用單例模式（Singleton），全域只存在兩個固定的實例：Py_True 與 Py_False。

是否有固定大小限制：有，因為直接對應共用的全域實例，大小完全固定。

如何擴充儲存空間：不需要擴充。

5. list（串列 / 向量）
是否為內建型別：是。

CPython 物件結構名稱：PyListObject。

如何儲存資料：內部維護一個指向其他 Python 物件指標的動態陣列（ob_item），以及記錄目前元素數（ob_size）與總配置容量（allocated）。

是否有固定大小限制：無固定的長度限制，可隨時新增或刪除元素。

如何擴充儲存空間：當串列空間用盡（ob_size == allocated）時，CPython 會自動配置一塊更大的記憶體空間（通常會預留額外空間以減少頻繁重新配置的開銷），並將舊的指標搬移過去。

6. set（集合）
是否為內建型別：是。

CPython 物件結構名稱：PySetObject。

如何儲存資料：採用雜湊表（Hash Table）結構，儲存指向鍵（key）的指標、雜湊值以及元素狀態。

是否有固定大小限制：無固定的元素數量限制，可隨元素增加而動態調整。

如何擴充儲存空間：當雜湊表的負載因子（load factor）超過特定門檻時，CPython 會自動建立一個更大的雜湊表，重新計算雜湊值並將現有元素遷移過去。

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？Python 採用了動態記憶體配置與大數處理機制，當數值超過一般硬體暫存器（如 64 位元）的範圍時，系統會自動擴充記憶體來容納更多的位數，因此不會發生傳統語言的整數溢位（overflow）現象
2. Python 的 `int` 是否具有固定的位元數限制？沒有。Python 的 int 並不像 C 語言的 int 那樣具有固定的 32 位元或 64 位元大小限制
3. 什麼是 Arbitrary Precision（任意精度）？Arbitrary Precision（任意精度）是指整數的精確度與大小可以隨著數值的增長而動態擴充，不受限於固定位元數，能夠完美、無失真地表示無限大的整數。
4. CPython 使用什麼物件來實作 `int`？在 CPython 原始碼實作中，Python 的 int 型別對應到底層的 C 結構是PyLongObject
5. Python 整數大小真正的限制是什麼？Python 整數大小真正且唯一的限制是電腦系統中可用的記憶體總量（RAM）。當整數大到將系統記憶體完全耗盡時，程式便會引發 MemoryError

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？因為 Python 的 float 在底層是透過 C 語言的雙精度浮點數（double）來實作，並遵循 IEEE 754 標準。電腦是用二進位（0與1）來儲存小數，而像 0.1（$1/10$）或 0.2 這種十進位小數，在二進位中會變成無限循環小數。電腦只能取近似值儲存，因此在進行加法運算時，累積的微小誤差會導致 0.1 + 0.2 變成 0.30000000000000004，進而使 0.1 + 0.2 == 0.3 判定為 False
2. 浮點數是否能精確表示所有小數？不行。 由於儲存位元數有限，只有那些剛好能被表示為「二進位有限小數」（例如 $0.5$、$0.25$ 等）的數字才能被精確表示，絕大多數的十進位小數都只能用近似值儲存
3. 什麼是浮點數的精度（precision）？精度是指浮點數能夠分辨兩個相鄰數值之間的「解析度」或「細膩程度」。以 IEEE 754 雙精度浮點數為例，它用 53 個位元來記錄有效數字，換算成十進位大約是 15 到 17 位有效數字
4. 什麼是浮點數的範圍（range）？範圍是指浮點數能夠表示的「最大值與最小值」（包含極大正數、極小正數等極端值）。這主要取決於指數部分的位元數。超出這個範圍就會發生「上溢（overflow，變成無限大 inf）」或「下溢（underflow，變成 0）」
5. 精度與範圍有何不同？精度（Precision）：決定了數字的精確細節（能保留幾位有效數字）。
範圍（Range）：決定了數字的極端大小（可以有多大或多小）

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？AI 模型中的權重（Weights）與偏差（Biases）通常使用浮點數型別（如 FP32、FP16、BF16）來儲存，因為神經網路在訓練時需要透過梯度下降法進行微小的數值調整。近年來為了節省記憶體與加速運算，也會使用低精度的整數或浮點數格式進行量化（Quantization），例如 INT8、INT4 甚至 FP8
2. 什麼是 FP64？全稱：64-bit Floating-Point（雙精度浮點數）。

特點：遵循 IEEE 754 標準，佔用 8 位元組（64 bits）的記憶體空間。它提供極高的精確度與極廣的數值範圍，但由於運算速度慢且記憶體消耗巨大，在深度學習中很少用作模型權重，主要用於高效能科學計算（HPC）。
3. 什麼是 FP32？全稱：32-bit Floating-Point（單精度浮點數）。

特點：遵循 IEEE 754 標準，佔用 4 位元組（32 bits）的記憶體空間（包含 1 個符號位、8 個指數位、23 個尾數位）。它是過去深度學習訓練與推理的標準精度，能夠提供足夠的精確度讓模型穩定收斂。
4. 什麼是 FP16？全稱：16-bit Floating-Point（半精度浮點數）。

特點：佔用 2 位元組（16 bits）的記憶體空間（1 個符號位、5 個指數位、10 個尾數位）。相比 FP32，它能節省一半的記憶體頻寬並加速運算，但因為動態範圍較小，極易發生「數值下溢（Underflow）」或「上溢（Overflow）」，通常需要搭配混合精度訓練（Mixed Precision Training）來使用。
5. 什麼是 BF16？全稱：Bfloat16（Brain Floating Point，由 Google Brain 提出）。

特點：同樣佔用 2 位元組（16 bits），但其配置方式與 FP16 不同：它保留了與 FP32 相同的 8 個指數位，僅將尾數縮減至 7 位。這讓 BF16 擁有與 FP32 完全相同的動態範圍，不容易發生數值溢位，因此在現代大型語言模型（LLM）的訓練與推理中被廣泛採用。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？記憶體開銷過大：Python 的內建 int（PyLongObject）在 CPython 中不是單純的原始資料，而是包含參考計數、型別指標等額外開銷的物件，動態長度機制會耗費大量記憶體。

無法發揮硬體加速：AI 模型的運算高度依賴 GPU 或 TPU 的矩陣運算單元（如 Tensor Cores），這些硬體專為固定大小的緊湊二進位格式（如 FP32、BF16、INT8）設計，無法高效處理 Python 的動態物件。

數值性質不符：神經網路的權重是介於連續區間的實數（小數），透過梯度下降不斷進行微小調整。若強行使用整數，不僅無法表達權重微調的精度，還會使反向傳播的優化過程完全失效。

## 第七部分：比較不同數值表示方式

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int |變動（動態擴充，無固定位元數）|無（整數運算完全精確）|一般程式設計、計數、索引、大整數運算|
| FP64 |64 bits (8 位元組)|有（但精確度極高，約 15-17 位有效數字）|高效能科學計算（HPC）、高精確度模擬|
| FP32 |32 bits (4 位元組) | 有（約 7 位有效數字） |傳統深度學習模型訓練與推理、標準科學計算 |
| FP16 |16 bits (2 位元組)|有（精確度較低，僅約 3-4 位有效數字，易溢位）|深度學習的混合精度訓練、加速運算與節省記憶體|
| BF16 |16 bits (2 位元組)|有（尾數較少，但動態範圍與 FP32 相同）|現代大型語言模型（LLM）的訓練與高效推理|

並回答：

1. 哪種表示方式最精確？int（整數）最精確，因為它具有任意精度（Arbitrary Precision），在進行整數運算時完全沒有誤差。若就實數/小數而言，則以 FP64 的精確度最高
2. 哪種表示方式最節省記憶體？FP16 與 BF16。兩者皆僅佔用 16 bits（2 位元組）的記憶體空間，是 FP32 的一半。
3. 哪種表示方式最常見於 AI 訓練？FP32 與 BF16。FP32 是傳統深度學習訓練的標準格式，而 BF16 則是現代大型語言模型訓練的主流選擇
4. 哪種表示方式最常見於 AI 推論（Inference）？FP32、BF16 以及 FP16。在實際部署與模型推理時，通常會根據效能與精度需求使用這些浮點數格式（或進一步量化）來加速運算

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？這代表整數的位元數與精確度不會被傳統程式語言的硬體限制（例如固定 32 位元或 64 位元）所綁死。整數的大小可以隨著數值的增長而動態調整，且在進行加減乘除等整數運算時，完全不會發生精度遺失或截斷誤差。
2. 它是否代表可以儲存無限大的整數？理論上可以，但在實務上不行。 雖然它的位元數沒有寫死的上限，但它仍然受到電腦硬體實際可用的記憶體總量（RAM）所限制。只要系統記憶體足夠，數值就能夠一直擴大；但若數值大到將記憶體完全耗盡，程式就會引發 MemoryError
3. CPython 如何實現這項特性？在 CPython 的底層實作中，整數是以 PyLongObject 結構來表示的。它使用一個動態調整大小的陣列（ob_digit）來將大整數拆解成多個數位區塊儲存，同時記錄其正負號與長度。當數字變大、現有記憶體空間不足時，CPython 會自動向系統申請更大的記憶體空間來容納更多的數位，藉此實現無限制的動態擴充

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？動態記憶體配置（Arbitrary Precision）：Python 的 int 在 CPython 底層是以 PyLongObject 結構來實作。當數值超過一般硬體暫存器（如 64 位元）的範圍時，系統會自動擴充內部陣列的記憶體空間來容納更多的位數。

只要電腦的系統記憶體（RAM）尚未耗盡，整數就能夠持續動態增長，因此在理論上沒有固定的上限。
2. 為什麼 `float` 卻存在範圍限制？固定的硬體結構：Python 的 float 對應到 C 語言的 double（雙精度浮點數），嚴格遵循 IEEE 754 標準。

它在記憶體中固定佔用 8 位元組（64 bits），其中分配給指數與尾數的位元數是固定的。因為記憶體空間大小不會隨數值大小改變，所以它能表示的極端數值範圍（最大值與最小值）是固定的。超出此範圍就會發生上溢（Overflow，變成 inf）或下溢（Underflow，變成 0）。
3. 為什麼 `float` 可能產生誤差？二進位無法完美表示所有十進位小數：電腦底層使用二進位（0 與 1）來運算，但許多常見的十進位小數（例如 0.1）在二進位中其實是無限循環小數。

有限的精確度（尾數截斷）：由於浮點數的有效位元（尾數）空間有限（雙精度約 15 到 17 位有效數字），電腦必須將無限循環小數進行捨去與近似。當這些帶有微小誤差的數值進行算術運算時，誤差便會累積放大，導致結果與數學上的精確答案產生微小的偏差（例如 0.1 + 0.2 不會精準等於 0.3）。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。正確性優先（Python int 的 Arbitrary Precision）：

核心目標：確保數學計算的絕對精確，絕不容許因硬體限制而發生截斷或溢位。

設計方式：採用動態記憶體配置，讓數值能夠隨著需求無限長大，位元數不固定。

效率優先（AI 模型的 FP16 / BF16）：

核心目標：追求極致的計算速度與最低的記憶體消耗。

設計方式：採用固定的 16 位元（2 位元組）低精度結構，牺牲部分數值精確度，以換取硬體的高效平行運算能力。
2. 各自的優點與缺點。Python int（Arbitrary Precision）

優點：完全沒有溢位問題，計算結果百分之百精確，適用於商業邏輯、精確計數與大數運算。

缺點：記憶體開銷大、運算速度慢，無法進行高密度的硬體平行加速。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？優點：佔用記憶體極少（僅 FP32 的一半），能大幅提升 GPU/TPU 的矩陣運算速度，並允許模型載入更大的參數規模。

缺點：存在精確度誤差，動態範圍有限（FP16 易發生溢位或下溢），可能影響模型收斂或推理精確度。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？記憶體需求暴增：FP64 佔用 64 位元（8 位元組），是 FP16/BF16 的 4 倍。這會導致模型大小暴增，原本可以在單張 GPU 上執行的模型，可能需要數十張甚至數百張 GPU 才能載入。

運算速度嚴重拖慢：硬體的 Tensor Core 主要針對低精度（如 FP16、BF16、INT8）進行硬體級加速。改用 FP64 會使運算吞吐量大幅下降，訓練與推理時間會慢上好幾倍，甚至無法在合理的時間內完成訓練。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
1. 記憶體使用量 (Memory Usage)
Python 大整數（10**1000）：
Python 的整數採用任意精度（Arbitrary Precision）設計，在 CPython 中透過 PyLongObject 實作。隨著數值位數增加，系統會動態配置更多記憶體來儲存數位。雖然能精確容納極大數字，但其物件結構本身包含額外的標頭與指標開銷，記憶體消耗隨位數增長。

AI 模型權重（FP32、FP16、BF16）：
AI 模型動輒包含數百億至數千億個參數，必須嚴格控制記憶體。FP32 固定佔用 4 位元組，而 FP16 與 BF16 更僅佔用 2 位元組。使用低精度浮點數可以大幅壓縮模型體積，讓龐大的模型能夠順利載入至 GPU 的 VRAM 中。

2. 計算速度 (Computation Speed)
Python 大整數：
由於位數不固定且可能非常巨大，當進行算術運算時，CPU 無法直接透過單一硬體指令完成，必須透過演算法模擬多位數運算（類似直式乘法），位數越多速度越慢。

AI 模型浮點數：
FP32、FP16 和 BF16 的位元數完全固定（分別為 32 或 16 位元）。現代 CPU 和 GPU 內部都有專為固定位元長度設計的算術邏輯單元（ALU），可以在極短的時脈週期內直接完成運算。

3. 數值精度 (Numerical Precision)
Python 大整數：
具備「正確性優先」的特性，整數運算具有無誤差（Exact）的絕對精度，絕不會因為空間不足而發生截斷。

AI 模型浮點數：
採用「效率優先」的有限精度設計。神經網路的權重與梯度本質上是連續的實數，且神經網路本身具備極高的容錯與自我修正能力。微小的捨入誤差（如 FP16 或 BF16 的精度限制）並不會影響模型的訓練收斂與推理表現。

4. GPU 運算效率 (GPU Computing Efficiency)
Python 大整數：
Python 的動態物件與大數機制設計複雜，無法被 GPU 的平行架構直接進行高效處理，若強行在 GPU 上運算會面臨嚴重的效能瓶頸。

AI 模型浮點數：
現代 GPU（如 NVIDIA 的 Tensor Cores）專為標準的低精度浮點數（FP32、FP16、BF16）設計了強大的硬體級矩陣乘法加速單元。使用這些格式能實現大規模的平行運算，將 AI 訓練與推理的速度提升數倍甚至數十倍。

總結來說：
Python 的 10**1000 追求的是數學邏輯的絕對正確與彈性；而 AI 模型選擇 FP32、FP16、BF16 則是為了在有限的硬體資源下，追求極致的記憶體效率與平行運算速度。
