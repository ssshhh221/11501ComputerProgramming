def get_zodiac():

    print("2. 民國年份 (例如: 115)")
    
    choice = input("請輸入選項 (1 或 2): ").strip()

    if choice == "1":
        try:
            gregorian_year = int(input("請輸入西元年份："))

    elif choice == "2":
        try:
            roc_year = int(input("請輸入民國年份："))
            # 民國轉西元
            gregorian_year = roc_year + 1911
        except ValueError:
            print("輸入錯誤，請輸入有效的數字！")
            return
    else:
        print("無效的選項，請重新執行程式。")
        return

    # 計算生肖索引（以 2004 年猴年為基準，2004 % 12 = 0）
    index = gregorian_year % 12
    animal = zodiacs[index]

    print("-" * 30)

if __name__ == "__main__":

Python 的 float 通常採用 IEEE 754 雙精度浮點格式 (double precision)，共 64 位元：

1 位元：符號位元 (Sign)
11 位元：指數 (Exponent)
52 位元：尾數 (Fraction)

其中尾數實際精度為 53 bits（含隱藏的最高位元）。
max：最大可表示的浮點數。
min：最小正規化浮點數。
dig：約可精確表示 15 位十進位有效數字。
mant_dig：尾數精度 53 bits。
epsilon：機器誤差，約 2.22×10⁻¹⁶。
radix：底數為 2（二進位）。
rounds：捨入模式，通常為 1，代表「四捨六入五成雙（Round Half Even）」。

因此 Python 的 float 精度約為 15~16 位有效數字，超過時可能產生誤差；若要處理無限多位數的年份，應使用 int（任意精度整數） 而非 float。

４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？

因為浮點數採用 IEEE 754 二進位表示，0.1 和 0.2 不能被精確儲存，所以產生微小誤差。可使用 round()、格式化輸出，或 Decimal 類別來解決。
5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。
我前面的方法其實只是「顯示上」或「特定情境下」的解決方案，不一定是科學或金融計算的好方法。

round() 是將結果四捨五入，可能掩蓋誤差，而不是消除誤差。
格式化輸出 "{:.1f}" 只是把誤差隱藏起來。
在大量運算後再一直 round()，反而可能累積更多誤差。
科學計算：使用 float，但以 math.isclose() 比較，不直接比較是否完全相等。
金融計算：使用 Decimal，避免 IEEE 754 浮點誤差。
round() 和格式化輸出只是修飾結果，不是根本解決方法。
